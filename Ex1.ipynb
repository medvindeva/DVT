```
import pandas as pd
import sqlite3

conn = sqlite3.connect("fifa.db")



csv_df = pd.read_csv("fifa_world_cup_2026_player_performance.csv")
csv_df.to_sql("player_performance", conn, if_exists="replace", index=False)
print("Ensured 'player_performance' table exists in fifa.db for reading.")


df = pd.read_sql("SELECT * FROM player_performance", conn)

print(df.head())

conn.close()
```

```
import pandas as pd
import requests
import io

url = "https://en.wikipedia.org/wiki/2026_FIFA_World_Cup"

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}


response = requests.get(url, headers=headers)


if response.status_code == 200:

    tables = pd.read_html(io.StringIO(response.text))
    print(tables[0].head())
else:
    print(f"Failed to retrieve page: Status Code {response.status_code}")
```
