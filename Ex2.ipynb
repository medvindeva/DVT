```python
import os
import pandas as pd
import sqlite3

# Dataset path (not strictly needed for direct file path, but kept for context)
path = "/content"

# Directly use the provided CSV file path
csv_file = "fifa_world_cup_2026_player_performance.csv"

print("CSV file:", csv_file)

# Load dataset
df = pd.read_csv(
    os.path.join(path, csv_file),
    low_memory=False
)

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset Information:")
df.info()

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nStatistical Description:")
display(df.describe().T)

print("\nMissing Values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nTeams:")
display(df["team"].value_counts().head(10))

print("\nPositions:")
display(df["position"].value_counts())

print("\nNationalities:")
display(df["nationality"].value_counts().head(10))

print("\nCorrelation:")
display(df.corr(numeric_only=True))

# Export CSV
df.to_csv(
    "fifa_player_performance_cleaned.csv",
    index=False
)

# Export Excel
df.to_excel(
    "fifa_player_performance.xlsx",
    index=False
)

# SQLite Database
conn = sqlite3.connect(
    "fifa_player_performance.db"
)

df.to_sql(
    "player_performance",
    conn,
    if_exists="replace",
    index=False
)

conn.close()

print(
    "\nCSV, Excel and SQLite database files "
    "created successfully."
)
```
