```python
import pandas as pd
import matplotlib.pyplot as plt

# Load FIFA World Cup 2026 Player Performance dataset
df = pd.read_csv("fifa_world_cup_2026_player_performance.csv")

# Display first five records
print("FIRST FIVE RECORDS")
print(df[['player_name', 'team', 'position',
          'goals', 'assists', 'player_rating']].head())

# Display shape
print("\nDATASET SHAPE")
print(df.shape)

# Display data types
print("\nDATA TYPES")
print(df[['age', 'goals', 'assists',
          'minutes_played', 'player_rating',
          'performance_score']].dtypes)

# Filtering players who scored more than 2 goals
print("\nPLAYERS WITH MORE THAN 2 GOALS")
high_goals = df[df["goals"] > 2]

print(high_goals[['player_name', 'team',
                   'goals', 'assists',
                   'player_rating']].head(10))

# Top 10 players based on player rating
print("\nTOP 10 PLAYERS BY PLAYER RATING")
top_players = df.nlargest(10, "player_rating")

print(top_players[['player_name', 'team',
                   'player_rating', 'goals',
                   'assists']])

# Statistical analysis
print("\nSTATISTICAL ANALYSIS")

print("Mean Goals =", df["goals"].mean())
print("Median Goals =", df["goals"].median())
print("Mean Assists =", df["assists"].mean())
print("Median Assists =", df["assists"].median())
print("Mean Minutes Played =", df["minutes_played"].mean())
print("Mean Player Rating =", df["player_rating"].mean())

# Grouping by position
print("\nAVERAGE PLAYER RATING BY POSITION")

position_rating = df.groupby("position")["player_rating"].mean()
print(position_rating)

# Correlation analysis
print("\nCORRELATION ANALYSIS")

corr = df[['goals',
           'assists',
           'shots',
           'minutes_played',
           'player_rating',
           'performance_score']].corr()

print(corr.round(2))

# Visualization
plt.figure(figsize=(8,5))
plt.scatter(df["minutes_played"], df["player_rating"], alpha=0.5)
plt.title("Minutes Played vs Player Rating")
plt.xlabel("Minutes Played")
plt.ylabel("Player Rating")
plt.show()
```
