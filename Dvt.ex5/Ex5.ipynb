```python
import pandas as pd
import matplotlib.pyplot as plt

# Load FIFA World Cup 2026 Player Performance dataset
df = pd.read_csv("fifa_world_cup_2026_player_performance.csv")

# 1. LINE CHART - Top 10 players by rating
top10 = df.nlargest(10, "player_rating")

plt.figure(figsize=(10,5))
plt.plot(top10["player_name"], top10["player_rating"], marker="o")
plt.title("Top 10 Players by Player Rating")
plt.xlabel("Player Name")
plt.ylabel("Player Rating")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 2. BAR CHART - Average rating by position
position_rating = df.groupby("position")["player_rating"].mean()

plt.figure(figsize=(8,5))
plt.bar(position_rating.index, position_rating.values)
plt.title("Average Player Rating by Position")
plt.xlabel("Position")
plt.ylabel("Average Player Rating")
plt.show()

# 3. PIE CHART - Player distribution by position
position_count = df["position"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(
    position_count.values,
    labels=position_count.index,
    autopct="%1.1f%%"
)
plt.title("Player Distribution by Position")
plt.show()

# 4. HISTOGRAM - Player rating distribution
plt.figure(figsize=(8,5))
plt.hist(df["player_rating"].dropna(), bins=10)
plt.title("Player Rating Distribution")
plt.xlabel("Player Rating")
plt.ylabel("Number of Players")
plt.show()

# 5. SCATTER PLOT - Minutes played vs rating
plt.figure(figsize=(8,5))
plt.scatter(
    df["minutes_played"],
    df["player_rating"],
    alpha=0.5
)
plt.title("Minutes Played vs Player Rating")
plt.xlabel("Minutes Played")
plt.ylabel("Player Rating")
plt.show()

# 6. BOX PLOT - Market value
plt.figure(figsize=(7,5))
plt.boxplot(df["market_value_eur"].dropna())
plt.title("Market Value Distribution")
plt.ylabel("Market Value (EUR)")
plt.show()

# 7. MULTIPLE LINE CHART - Goals and assists
top_players = df.nlargest(10, "player_rating")

plt.figure(figsize=(10,5))
plt.plot(
    top_players["player_name"],
    top_players["goals"],
    marker="o",
    label="Goals"
)
plt.plot(
    top_players["player_name"],
    top_players["assists"],
    marker="o",
    label="Assists"
)
plt.title("Goals and Assists of Top 10 Rated Players")
plt.xlabel("Player Name")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.tight_layout()
plt.show()
```
